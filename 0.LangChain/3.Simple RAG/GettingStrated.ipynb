```python
#Step 5 load PDF

from langchain_community.document_loaders import PyPDFLoader

# Option 1: Use raw string (recommended)
#loader = PyPDFLoader(r"C:\DS2026\Python\DataScience2026\10.LangChain\3.Simple RAG\Data\LLM.pdf")

# Option 2: Escape backslashes
#loader = PyPDFLoader("C:\\DS2026\\Python\\DataScience2026\\10.LangChain\\3.Simple RAG\\Data\\LLM.pdf")

# Option 3: Use forward slashes (works on Windows too)
loader = PyPDFLoader("C:/DS2026/Python/DataScience2026/10.LangChain/3.Simple RAG/Data/LLMs.pdf")


docs = loader.load()
```

```python
#Step 6 — Split Documents
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(

    chunk_size=1000,

    chunk_overlap=200

)

chunks = splitter.split_documents(docs)
```

```python
#Step 7 — Create Embeddings

#Using Ollama embeddings:

from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
#Step 8 — Store in ChromaDB
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(

    chunks,

    embedding=embeddings,

    persist_directory="chroma_db"

)
```

```python
from langchain_chroma import Chroma
```

```python
#Step 9 — Load Existing Database


db = Chroma(
    persist_directory="chroma_db",
    embedding_function=embeddings
)
```

```python
#Step 10 — Create Retriever
retriever = db.as_retriever(
    search_kwargs={"k":3}
)
```

```python
#Step 11 — Ask Question
question = "What is Machine Learning?"

docs = retriever.invoke(question)
```

```python
print("Number of retrieved chunks:", len(docs))

for i, doc in enumerate(docs):
    print(f"\nChunk {i+1}")
    print("-" * 50)
    print(doc.page_content)
```

```python
#Step 12 — Prompt Template
from langchain_core.prompts import PromptTemplate

template = """
Answer only from the context.

Context:
{context}

Question:
{question}
"""

prompt = PromptTemplate.from_template(template)
```

```python
context = "\n\n".join([doc.page_content for doc in docs])

formatted_prompt = prompt.format(
    context=context,
    question=question
)

print(formatted_prompt)
```

```python
#Step 13: Send the prompt to Llama

#Now call the model:

from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2")

response = llm.invoke(formatted_prompt)

print(response.content)
```
