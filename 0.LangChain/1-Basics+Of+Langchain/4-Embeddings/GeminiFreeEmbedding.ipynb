```python
import os
from dotenv import load_dotenv
load_dotenv()  # load environment variables
```

```python
# Set Gemini API key (from Google AI Studio or Vertex AI)
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
# Use Gemini embeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Use Gemini free embedding model
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")

text = "This is a tutorial on Gemini embedding"
query_result = embeddings.embed_query(text)

print(len(query_result))   # vector length
print(query_result[:10])   # preview first 10 values
```

```python
# Example: single query embedding
text = "This is a tutorial on Gemini embedding"
query_result = embeddings.embed_query(text)
print(len(query_result))   # vector length
print(query_result[:10])   # preview first 10 values
```

```python
# Example: embeddings with custom dimension (Matryoshka Representation Learning)
embeddings_512 = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    dimensions=512
)
```

```python
query_result = embeddings_512.embed_query(text)
print(len(query_result))
print(query_result[:10])
```
