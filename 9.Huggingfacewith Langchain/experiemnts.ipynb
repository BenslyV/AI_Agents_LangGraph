### Hugging Face x LangChain : A new partner package in LangChain
langchain_huggingface, a partner package in LangChain jointly maintained by Hugging Face and LangChain. This new Python package is designed to bring the power of the latest development of Hugging Face into LangChain and keep it up to date.

```python
import os
from dotenv import load_dotenv

from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
```

```python
load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.2-1B-Instruct",
    max_new_tokens=150,
    temperature=0.7,
    huggingfacehub_api_token=os.getenv("HF_TOKEN"),
)
```

```python
chat_model = ChatHuggingFace(llm=llm)

response = chat_model.invoke(
    "What is machine learning?"
)

print(response.content)
```

```python
response = chat_model.invoke(
    "What is Generative AI?"
)

print(response)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
template = """
Question: {question}

Answer: Let's think step by step.
"""

prompt = PromptTemplate.from_template(template)

print(prompt)
```

```python
formatted_prompt = prompt.invoke(
    {"question": "What is machine learning?"}
)

print(formatted_prompt)
```

```python
llm_chain = prompt | chat_model
```

```python
response = llm_chain.invoke(
    {"question": "What is machine learning?"}
)

print(response)
```

```python
template = """
You are a helpful AI assistant.

Question:
{question}

Provide a simple and clear answer.
"""

prompt = PromptTemplate.from_template(template)

chain = prompt | chat_model
```

```python
response = chain.invoke(
    {"question": "What is Generative AI?"}
)

print(response)
```

```python
#Hugging Face Embeddings

from langchain_huggingface import HuggingFaceEmbeddings
```

```python
model_name = "BAAI/bge-small-en"

model_kwargs = {
    "device": "cpu"
}

encode_kwargs = {
    "normalize_embeddings": True
}

hf_embeddings = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)
```

```python
embedding = hf_embeddings.embed_query(
    "hi this is harrison"
)

print(embedding)
```

```python
embedding = hf_embeddings.embed_query(
    "hi this is harrison"
)

print("Embedding type:", type(embedding))
print("Embedding dimensions:", len(embedding))
```

```python
texts = [
    "I love machine learning.",
    "Artificial intelligence is interesting.",
    "I like eating pizza."
]

embeddings = hf_embeddings.embed_documents(texts)

print("Number of documents:", len(embeddings))
print("Dimensions:", len(embeddings[0]))
```
